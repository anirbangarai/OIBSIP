# Data Cleaning

Before performing Exploratory Data Analysis (EDA), we need to clean the dataset.

The following preprocessing steps will be performed:

- Check Missing Values
- Check Duplicate Records
- Remove Duplicate Records (if any)
- Convert Date Column into Datetime Format
- Create New Features (Month & Year)

```python
# Checking Missing Values

df.isnull().sum()
```

# Missing Values Analysis

Missing values can negatively affect the analysis.

Let's visualize missing values using a heatmap.

```python
plt.figure(figsize=(10,5))

sns.heatmap(df.isnull(), cbar=False, cmap='viridis')

plt.title("Missing Values Heatmap")

plt.show()
```

### Observation

- The heatmap clearly highlights missing values.
- Missing values, if present, should be handled before analysis.

# Duplicate Records

Duplicate records unnecessarily increase dataset size and may affect analysis.

Let's check whether duplicates exist.

```python
print("Duplicate Rows :", df.duplicated().sum())
```

# Removing Duplicate Records

```python
df = df.drop_duplicates()

print("New Shape :", df.shape)
```

# Date Conversion

Currently, the Date column is stored as text.

We will convert it into datetime format for time-series analysis.

```python
df['Date'] = pd.to_datetime(df['Date'])
```

# Verify Date Conversion

```python
df.dtypes
```

# Creating Month and Year Columns

Extracting Month and Year helps us analyze unemployment trends over time.

```python
df['Month'] = df['Date'].dt.month_name()

df['Year'] = df['Date'].dt.year

df.head()
```

# Unique Regions

```python
df['Region'].unique()
```

# Number of States

```python
print("Total Regions :",df['Region'].nunique())
```

# Area Distribution

The dataset contains both Urban and Rural unemployment records.

Let's check their distribution.

```python
df['Area'].value_counts()
```

# Exploratory Data Analysis (EDA)

EDA helps us understand the overall characteristics of the dataset through visualizations.

# Distribution of Area

```python
plt.figure(figsize=(7,5))

sns.countplot(data=df,x='Area',palette='Set2')

plt.title("Urban vs Rural Records")

plt.xlabel("Area")

plt.ylabel("Count")

plt.show()
```

### Observation

- The chart shows the distribution of Urban and Rural observations.
- This helps understand how balanced the dataset is.

# Distribution of Unemployment Rate

```python
plt.figure(figsize=(10,6))

sns.histplot(df['Estimated Unemployment Rate (%)'],
             bins=25,
             kde=True,
             color='royalblue')

plt.title("Distribution of Estimated Unemployment Rate")

plt.xlabel("Unemployment Rate (%)")

plt.ylabel("Frequency")

plt.show()
```

### Observation

- The histogram shows how unemployment rates are distributed.
- The KDE curve provides a smooth estimate of the data distribution.

# Box Plot of Unemployment Rate

Box plots help detect outliers in the dataset.

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    x=df['Estimated Unemployment Rate (%)'],
    color='orange'
)

plt.title("Box Plot of Estimated Unemployment Rate")

plt.show()
```

### Observation

- Any points outside the whiskers represent potential outliers.
- Outliers may indicate unusually high unemployment in some regions.

# Average Unemployment Rate by Area

```python
area_avg = df.groupby('Area')['Estimated Unemployment Rate (%)'].mean()

area_avg
```

# Visualizing Average Unemployment Rate by Area

```python
plt.figure(figsize=(7,5))

area_avg.plot(kind='bar')

plt.title("Average Unemployment Rate by Area")

plt.xlabel("Area")

plt.ylabel("Average Unemployment Rate")

plt.xticks(rotation=0)

plt.show()
```

### Observation

- This graph compares the average unemployment rate in Urban and Rural areas.
- Differences may indicate varying economic impacts across regions.

---

# Month-wise Average Unemployment Trend

Time-series analysis helps us understand how unemployment changes over different months.

We calculate the average unemployment rate for each month.

```python
month_order = [
    'January','February','March','April',
    'May','June','July','August',
    'September','October','November','December'
]

monthly_avg = df.groupby('Month')['Estimated Unemployment Rate (%)'].mean()

monthly_avg = monthly_avg.reindex(month_order)

monthly_avg
```

# Visualizing Monthly Average Unemployment

```python
plt.figure(figsize=(12,6))

sns.lineplot(
    x=monthly_avg.index,
    y=monthly_avg.values,
    marker='o',
    linewidth=3
)

plt.title("Average Monthly Unemployment Rate")

plt.xlabel("Month")

plt.ylabel("Average Unemployment Rate (%)")

plt.xticks(rotation=45)

plt.grid(True)

plt.show()
```

### Observation

- The unemployment rate changes throughout the year.
- Some months experienced noticeably higher unemployment than others.
- Seasonal and economic factors may influence these fluctuations.

# Top 10 States with Highest Average Unemployment

Identifying the states with the highest unemployment helps understand regional economic challenges.

```python
top_states = df.groupby('Region')['Estimated Unemployment Rate (%)'] \
               .mean() \
               .sort_values(ascending=False) \
               .head(10)

top_states
```

# Visualizing Top 10 States

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x=top_states.values,
    y=top_states.index,
    palette='Reds_r'
)

plt.title("Top 10 States by Average Unemployment Rate")

plt.xlabel("Average Unemployment Rate (%)")

plt.ylabel("State")

plt.show()
```

### Observation

- These states recorded the highest average unemployment rates.
- Policymakers can focus on these regions for employment initiatives.

# Top 10 States with Highest Employment

```python
top_employed = df.groupby('Region')['Estimated Employed'] \
                 .mean() \
                 .sort_values(ascending=False) \
                 .head(10)

top_employed
```

# Visualizing Top Employment States

```python
plt.figure(figsize=(12,6))

sns.barplot(
    x=top_employed.values,
    y=top_employed.index,
    palette='Greens_r'
)

plt.title("Top 10 States by Average Employment")

plt.xlabel("Estimated Employed")

plt.ylabel("State")

plt.show()
```

### Observation

- These states have the highest average employment levels.
- Larger populations generally correspond to higher employment numbers.

# Labour Participation Rate Across States

```python
labour_rate = df.groupby('Region')['Estimated Labour Participation Rate (%)'] \
                .mean() \
                .sort_values(ascending=False)

labour_rate.head()
```

# Visualizing Labour Participation Rate

```python
plt.figure(figsize=(14,8))

sns.barplot(
    x=labour_rate.values,
    y=labour_rate.index,
    palette='viridis'
)

plt.title("Average Labour Participation Rate by State")

plt.xlabel("Labour Participation Rate (%)")

plt.ylabel("State")

plt.show()
```

### Observation

- Labour participation differs significantly across states.
- A higher participation rate indicates a larger share of the working-age population is active in the labour force.

# Correlation Analysis

Correlation measures the relationship between numerical variables.

```python
correlation = df.select_dtypes(include=['number']).corr()

correlation
```

# Correlation Heatmap

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)

plt.title("Correlation Heatmap")

plt.show()
```

### Observation

- Positive values indicate a direct relationship.
- Negative values indicate an inverse relationship.
- Strong correlations help identify variables that influence unemployment.

# Scatter Plot

Let's examine the relationship between employment and unemployment.

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x='Estimated Employed',
    y='Estimated Unemployment Rate (%)',
    hue='Area'
)

plt.title("Employment vs Unemployment Rate")

plt.show()
```

### Observation

- The scatter plot illustrates the relationship between employment and unemployment.

# Pair Plot

A pair plot helps visualize relationships among multiple numerical variables simultaneously.

```python
sns.pairplot(
    df[
        [
            'Estimated Unemployment Rate (%)',
            'Estimated Employed',
            'Estimated Labour Participation Rate (%)'
        ]
    ]
)

plt.show()
```

# COVID-19 Impact Analysis

The COVID-19 pandemic significantly affected employment across India.

To understand its impact, we compare the average unemployment rate before and after March 2020.

```python
# Create COVID period column

df['COVID Period'] = df['Date'].apply(
    lambda x: 'Before COVID' if x < pd.Timestamp('2020-03-01') else 'During/After COVID'
)

df[['Date', 'COVID Period']].head()
```

# Average Unemployment Before and During/After COVID

```python
covid_avg = df.groupby('COVID Period')['Estimated Unemployment Rate (%)'].mean()

covid_avg
```

# Visualizing COVID Impact

```python
plt.figure(figsize=(7,5))

sns.barplot(
    x=covid_avg.index,
    y=covid_avg.values,
    palette='Set1'
)

plt.title("Average Unemployment Rate: Before vs During/After COVID")

plt.xlabel("Period")

plt.ylabel("Average Unemployment Rate (%)")

plt.show()
```

### Observation

- The chart highlights the impact of the COVID-19 pandemic on unemployment.
- A higher unemployment rate during/after COVID reflects the disruption caused by lockdowns and reduced economic activity.

```python
# Average Unemployment Rate by Year
```

```python
year_avg = df.groupby('Year')['Estimated Unemployment Rate (%)'].mean()

year_avg
```

# Visualizing Year-wise Unemployment

```python
plt.figure(figsize=(8,5))

year_avg.plot(
    kind='bar',
    color='steelblue'
)

plt.title("Average Unemployment Rate by Year")

plt.xlabel("Year")

plt.ylabel("Average Unemployment Rate (%)")

plt.xticks(rotation=0)

plt.show()
```

### Observation

- Comparing yearly averages helps identify long-term changes in unemployment.
- The rise in 2020 reflects the economic effects of the pandemic.
